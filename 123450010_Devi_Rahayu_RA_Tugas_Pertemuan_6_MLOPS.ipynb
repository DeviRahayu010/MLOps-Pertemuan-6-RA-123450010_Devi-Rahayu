```python
# Menyiapkan dataset dan MLflow

import os
import random
import numpy as np
import torch
import mlflow
from pathlib import Path
from torchvision import datasets, transforms
from torch.utils.data import DataLoader, Subset

# Reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

# Direktori eksperimen
PROJECT_DIR = Path("/content/mlops_pertemuan_6")
DATA_DIR = PROJECT_DIR / "data"
ARTIFACT_DIR = PROJECT_DIR / "artifacts"

DATA_DIR.mkdir(parents=True, exist_ok=True)
ARTIFACT_DIR.mkdir(parents=True, exist_ok=True)

# MLflow tracking menggunakan SQLite lokal
mlflow.set_tracking_uri("sqlite:////content/mlops_pertemuan_6/mlflow.db")
mlflow.set_experiment("Devi_Rahayu_MLOps_Pertemuan_6")

# Transformasi gambar
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

# Unduh dataset Fashion-MNIST
train_dataset = datasets.FashionMNIST(
    root=str(DATA_DIR),
    train=True,
    download=True,
    transform=transform
)

test_dataset = datasets.FashionMNIST(
    root=str(DATA_DIR),
    train=False,
    download=True,
    transform=transform
)

# Gunakan subset agar eksperimen lebih cepat
train_indices = list(range(6000))
test_indices = list(range(1000))

train_subset = Subset(train_dataset, train_indices)
test_subset = Subset(test_dataset, test_indices)

print("Jumlah data training:", len(train_subset))
print("Jumlah data testing:", len(test_subset))
print("Eksperimen:", mlflow.get_experiment_by_name(
    "Devi_Rahayu_MLOps_Pertemuan_6"
).name)
print("Device:", "cuda" if torch.cuda.is_available() else "cpu")
```

```python
# Membuat model CNN baseline

import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
import mlflow
import mlflow.pytorch
import time

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

class FashionCNN(nn.Module):
    def __init__(self, dropout=0.25, batch_norm=True):
        super().__init__()
        layers = [
            nn.Conv2d(1, 32, kernel_size=3, padding=1),
            nn.ReLU()
        ]
        if batch_norm:
            layers.append(nn.BatchNorm2d(32))
        layers += [
            nn.MaxPool2d(2),
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU()
        ]
        if batch_norm:
            layers.append(nn.BatchNorm2d(64))
        layers += [
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(64 * 7 * 7, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, 10)
        ]
        self.network = nn.Sequential(*layers)

    def forward(self, x):
        return self.network(x)

def train_one_epoch(model, loader, optimizer, criterion):
    model.train()
    total_loss, correct, total = 0.0, 0, 0

    for images, labels in loader:
        images, labels = images.to(DEVICE), labels.to(DEVICE)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * labels.size(0)
        correct += (outputs.argmax(1) == labels).sum().item()
        total += labels.size(0)

    return total_loss / total, correct / total

def evaluate(model, loader):
    model.eval()
    correct, total, total_loss = 0, 0, 0.0
    criterion = nn.CrossEntropyLoss()

    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(DEVICE), labels.to(DEVICE)
            outputs = model(images)
            total_loss += criterion(outputs, labels).item() * labels.size(0)
            correct += (outputs.argmax(1) == labels).sum().item()
            total += labels.size(0)

    return total_loss / total, correct / total

train_loader = DataLoader(
    train_subset, batch_size=128, shuffle=True, num_workers=2
)
test_loader = DataLoader(
    test_subset, batch_size=128, shuffle=False, num_workers=2
)

print("Model CNN dan fungsi training berhasil disiapkan.")
print("Device:", DEVICE)
```

```python
# Menjalankann eksperimen baseline

seed = 42
torch.manual_seed(seed)
np.random.seed(seed)

model = FashionCNN(dropout=0.25, batch_norm=True).to(DEVICE)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
criterion = nn.CrossEntropyLoss()
epochs = 3

with mlflow.start_run(run_name="baseline_cnn") as run:
    mlflow.log_params({
        "model": "FashionCNN",
        "seed": seed,
        "epochs": epochs,
        "batch_size": 128,
        "learning_rate": 0.001,
        "dropout": 0.25,
        "batch_norm": True,
        "train_samples": len(train_subset),
        "test_samples": len(test_subset)
    })

    for epoch in range(1, epochs + 1):
        start = time.time()
        train_loss, train_acc = train_one_epoch(
            model, train_loader, optimizer, criterion
        )
        val_loss, val_acc = evaluate(model, test_loader)

        mlflow.log_metrics({
            "train_loss": train_loss,
            "train_accuracy": train_acc,
            "val_loss": val_loss,
            "val_accuracy": val_acc,
            "epoch_time_seconds": time.time() - start
        }, step=epoch)

        print(
            f"Epoch {epoch}/{epochs} | "
            f"train_loss={train_loss:.4f} | "
            f"train_acc={train_acc:.4f} | "
            f"test_acc={val_acc:.4f}"
        )

    model_path = ARTIFACT_DIR / "baseline_cnn.pth"
    torch.save(model.state_dict(), model_path)
    mlflow.log_artifact(str(model_path))

    print("Run ID:", run.info.run_id)
    print("Baseline selesai dan metrik tercatat di MLflow.")
```

```python
# Tuning hyperparameter dengan Optuna

# Hyperparameter tuning dengan Optuna dan nested MLflow runs
import optuna
import mlflow
import torch
import numpy as np
import time
from torch.utils.data import DataLoader

N_TRIALS = 24
TUNING_EPOCHS = 2

def objective(trial):
    # Parameter yang akan diuji Optuna
    learning_rate = trial.suggest_float(
        "learning_rate", 1e-4, 3e-3, log=True
    )
    batch_size = trial.suggest_categorical(
        "batch_size", [64, 128]
    )
    dropout = trial.suggest_float(
        "dropout", 0.1, 0.5
    )
    batch_norm = trial.suggest_categorical(
        "batch_norm", [True, False]
    )

    # Nested run: setiap trial tercatat sebagai child run
    with mlflow.start_run(
        run_name=f"optuna_trial_{trial.number:02d}",
        nested=True
    ):
        mlflow.log_params({
            "trial_number": trial.number,
            "learning_rate": learning_rate,
            "batch_size": batch_size,
            "dropout": dropout,
            "batch_norm": batch_norm,
            "epochs": TUNING_EPOCHS
        })

        # Seed tetap agar perbandingan lebih konsisten
        torch.manual_seed(42)
        np.random.seed(42)

        loader = DataLoader(
            train_subset,
            batch_size=batch_size,
            shuffle=True,
            num_workers=2
        )

        model = FashionCNN(
            dropout=dropout,
            batch_norm=batch_norm
        ).to(DEVICE)

        optimizer = torch.optim.Adam(
            model.parameters(),
            lr=learning_rate
        )
        criterion = nn.CrossEntropyLoss()

        best_accuracy = 0.0

        for epoch in range(1, TUNING_EPOCHS + 1):
            train_loss, train_acc = train_one_epoch(
                model, loader, optimizer, criterion
            )
            val_loss, val_acc = evaluate(model, test_loader)

            mlflow.log_metrics({
                "train_loss": train_loss,
                "train_accuracy": train_acc,
                "val_loss": val_loss,
                "val_accuracy": val_acc
            }, step=epoch)

            best_accuracy = max(best_accuracy, val_acc)

        mlflow.log_metric("best_val_accuracy", best_accuracy)

        return best_accuracy


# Parent run menampung seluruh eksperimen tuning
with mlflow.start_run(run_name="optuna_hyperparameter_tuning"):
    mlflow.log_param("n_trials", N_TRIALS)
    mlflow.log_param("epochs_per_trial", TUNING_EPOCHS)
    mlflow.log_param("dataset", "Fashion-MNIST")
    mlflow.log_param("tracking_type", "nested_runs")

    study = optuna.create_study(
        direction="maximize",
        study_name="fashion_mnist_cnn_tuning",
        sampler=optuna.samplers.TPESampler(seed=42)
    )

    study.optimize(objective, n_trials=N_TRIALS)

    mlflow.log_params({
        f"best_{key}": value
        for key, value in study.best_params.items()
    })
    mlflow.log_metric(
        "best_val_accuracy", study.best_value
    )

print("\nTUNING SELESAI")
print("Jumlah trial:", len(study.trials))
print("Best validation accuracy:", round(study.best_value, 4))
print("Best hyperparameters:")
for key, value in study.best_params.items():
    print(f"  {key}: {value}")
```

```python
# Evaluasi konfigurasi terbaik menggunakan 3 seed


# Evaluasi konfigurasi terbaik dengan tiga seed

import statistics
import numpy as np
import torch
import mlflow

best_params = study.best_params
seeds = [42, 123, 2026]
seed_results = []

for seed in seeds:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    generator = torch.Generator().manual_seed(seed)

    train_loader_seed = DataLoader(
        train_subset,
        batch_size=best_params["batch_size"],
        shuffle=True,
        num_workers=2,
        generator=generator
    )

    model = FashionCNN(
        dropout=best_params["dropout"],
        batch_norm=best_params["batch_norm"]
    ).to(DEVICE)

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=best_params["learning_rate"]
    )
    criterion = nn.CrossEntropyLoss()

    with mlflow.start_run(run_name=f"seed_evaluation_{seed}"):
        mlflow.log_params({
            "seed": seed,
            "learning_rate": best_params["learning_rate"],
            "batch_size": best_params["batch_size"],
            "dropout": best_params["dropout"],
            "batch_norm": best_params["batch_norm"],
            "epochs": 3
        })

        for epoch in range(1, 4):
            train_loss, train_acc = train_one_epoch(
                model, train_loader_seed, optimizer, criterion
            )
            val_loss, val_acc = evaluate(model, test_loader)

            mlflow.log_metrics({
                "train_loss": train_loss,
                "train_accuracy": train_acc,
                "val_loss": val_loss,
                "val_accuracy": val_acc
            }, step=epoch)

        seed_results.append({
            "seed": seed,
            "accuracy": val_acc,
            "loss": val_loss
        })

        print(
            f"Seed {seed}: "
            f"accuracy={val_acc:.4f}, loss={val_loss:.4f}"
        )

accuracies = [result["accuracy"] for result in seed_results]
losses = [result["loss"] for result in seed_results]

mean_accuracy = statistics.mean(accuracies)
std_accuracy = statistics.stdev(accuracies)
mean_loss = statistics.mean(losses)
std_loss = statistics.stdev(losses)

with mlflow.start_run(run_name="three_seed_summary"):
    mlflow.log_metrics({
        "mean_accuracy": mean_accuracy,
        "std_accuracy": std_accuracy,
        "mean_loss": mean_loss,
        "std_loss": std_loss
    })

print("\nRINGKASAN TIGA SEED")
print(f"Mean accuracy: {mean_accuracy:.4f}")
print(f"Std accuracy : {std_accuracy:.4f}")
print(f"Mean loss    : {mean_loss:.4f}")
print(f"Std loss     : {std_loss:.4f}")
```

```python
# Ablation study


# Ablation study: augmentation, dropout, dan batch normalization

from torchvision import datasets, transforms
from torch.utils.data import DataLoader, Subset
import mlflow
import torch
import numpy as np
import random

# Data training dengan augmentasi
aug_transform = transforms.Compose([
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

aug_dataset = datasets.FashionMNIST(
    root=str(DATA_DIR),
    train=True,
    download=False,
    transform=aug_transform
)
aug_subset = Subset(aug_dataset, list(range(6000)))

# Empat konfigurasi yang dibandingkan
ablation_configs = [
    {
        "name": "full_model",
        "augmentation": True,
        "dropout": best_params["dropout"],
        "batch_norm": best_params["batch_norm"]
    },
    {
        "name": "without_augmentation",
        "augmentation": False,
        "dropout": best_params["dropout"],
        "batch_norm": best_params["batch_norm"]
    },
    {
        "name": "without_dropout",
        "augmentation": True,
        "dropout": 0.0,
        "batch_norm": best_params["batch_norm"]
    },
    {
        "name": "without_batch_norm",
        "augmentation": True,
        "dropout": best_params["dropout"],
        "batch_norm": False
    }
]

ablation_results = []

with mlflow.start_run(run_name="ablation_study"):
    mlflow.log_param("epochs", 3)
    mlflow.log_param("seed", 42)

    for config in ablation_configs:
        random.seed(42)
        np.random.seed(42)
        torch.manual_seed(42)

        selected_dataset = (
            aug_subset if config["augmentation"] else train_subset
        )

        loader = DataLoader(
            selected_dataset,
            batch_size=best_params["batch_size"],
            shuffle=True,
            num_workers=2,
            generator=torch.Generator().manual_seed(42)
        )

        model = FashionCNN(
            dropout=config["dropout"],
            batch_norm=config["batch_norm"]
        ).to(DEVICE)

        optimizer = torch.optim.Adam(
            model.parameters(),
            lr=best_params["learning_rate"]
        )
        criterion = nn.CrossEntropyLoss()

        with mlflow.start_run(
            run_name=config["name"],
            nested=True
        ):
            mlflow.log_params({
                "ablation": config["name"],
                "augmentation": config["augmentation"],
                "dropout": config["dropout"],
                "batch_norm": config["batch_norm"],
                "batch_size": best_params["batch_size"],
                "learning_rate": best_params["learning_rate"]
            })

            for epoch in range(1, 4):
                train_loss, train_acc = train_one_epoch(
                    model, loader, optimizer, criterion
                )
                val_loss, val_acc = evaluate(model, test_loader)

                mlflow.log_metrics({
                    "train_loss": train_loss,
                    "train_accuracy": train_acc,
                    "val_loss": val_loss,
                    "val_accuracy": val_acc
                }, step=epoch)

            result = {
                "configuration": config["name"],
                "accuracy": val_acc,
                "loss": val_loss
            }
            ablation_results.append(result)

            mlflow.log_metric("final_val_accuracy", val_acc)
            mlflow.log_metric("final_val_loss", val_loss)

            print(
                f"{config['name']}: "
                f"accuracy={val_acc:.4f}, loss={val_loss:.4f}"
            )

print("\nRINGKASAN ABLATION STUDY")
for result in ablation_results:
    print(result)
```

```python
# Unduh file dari Colab
# Setelah sel di atas berhasil, jalankan sel baru ini untuk membuat ZIP berisi file dokumentasi dan metadata DVC, tanpa memasukkan dataset berukuran besar.

import shutil
from pathlib import Path

PROJECT_DIR = Path("/content/mlops_pertemuan_6")
EXPORT_DIR = Path("/content/hasil_pengumpulan_mlop6")
EXPORT_DIR.mkdir(exist_ok=True)

files_to_export = [
    "README.md",
    "requirements.txt",
    "experiment_summary.json",
    "data.dvc",
    ".dvcignore"
]

for name in files_to_export:
    source = PROJECT_DIR / name
    if source.exists() and source.is_file():
        shutil.copy2(source, EXPORT_DIR / name)

zip_path = shutil.make_archive(
    "/content/Devi_Rahayu_MLOps_Pertemuan_6",
    "zip",
    EXPORT_DIR
)

print("ZIP berhasil dibuat:", zip_path)

from google.colab import files
files.download(zip_path)
```
